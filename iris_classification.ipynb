# Лабораторна робота №4. Класифікація Iris


## Завантаження даних та підготовка

```
from main import load_data, split_data, LABELS
X, y = load_data()
X_train, X_test, y_train, y_test = split_data(X, y)
print('Усього:', len(X), 'Навчання:', len(X_train), 'Тест:', len(X_test))
print('Класи:', LABELS)
```

##  Алгоритм KNN

```
from main import KNNClassifier, euclidean_distance
model = KNNClassifier(k=1)
model.fit(X_train, y_train)
print('Відстань між двома першими квітками:', euclidean_distance(X[0], X[1]))
```

## Прогноз нового ірису

```
new_iris = [5.0, 2.9, 1.0, 0.2]
print('Прогноз:', model.predict_one(new_iris))
```

## Оцінка на тестовій вибірці

```
from main import confusion_matrix
predictions = model.predict(X_test)
print('Accuracy:', round(model.score(X_test, y_test), 4))
print('Матриця помилок:')
for label, row in zip(LABELS, confusion_matrix(y_test, predictions)):
    print(label, row)
```
